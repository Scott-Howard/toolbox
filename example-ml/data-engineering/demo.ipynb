```python
# sample data generation

import pandas as pd
import numpy as np
import json
from datetime import datetime, timedelta

#create streaming event
np.random.seed(42)
events = []
device = ['mobile','desktop','tablet']
ppath = ['/home','/products','/checkout']

for i in range(1000):
    properties = {
        'device_type': np.random.choice(device),
        'page_path': np.random.choice(ppath),
        'session_length': np.random.randint(60,3600)
    }
    if np.random.random() >0/7:
        properties['purchase_value'] = round(np.random.uniform(20,300),2)
    event = {
        'event_id': f'evt_{i}',
        'time_stamp': (datetime.now()-timedelta(hours=np.random.randint(0,96))).isoformat(),
        'user_id': f'user_{np.random.randint(100,999)}',
        'event_type': np.random.choice(['view','click','purchase']),
        'metadata': json.dumps(properties)
    }
    events.append(event)

# create performance log
db_logs = pd.DataFrame({
    'timestamp': pd.date_range('2024-01-01', periods=500, freq='1min'),
    ''
})
```
